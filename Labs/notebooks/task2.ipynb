```python
from pathlib import Path
import sys

current_dir = Path.cwd().resolve()
if (current_dir / "src").exists():
    project_root = current_dir
elif (current_dir.parent / "src").exists():
    project_root = current_dir.parent
else:
    project_root = current_dir

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.inference import inferencia_marginal

prior_alarm = inferencia_marginal({"a": 1}, {})
diagnostic = inferencia_marginal({"b": 1}, {"a": 1})
explain_away = inferencia_marginal({"b": 1}, {"a": 1, "e": 1})

print(f"P(A = 1) = {prior_alarm:.4f}")
print(f"P(B = 1 | A = 1) = {diagnostic:.4f}")
print(f"P(B = 1 | A = 1, E = 1) = {explain_away:.4f}")
print()
print(
    "Conclusion: al observar la alarma, la hipotesis de robo gana peso; "
    "pero cuando se confirma el terremoto, la necesidad de explicar la alarma "
    "con robo disminuye de forma clara. Eso demuestra el efecto explain away."
)
```
